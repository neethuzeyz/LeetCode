**TASK: Use requests library and BeautifulSoup to connect to http://quotes.toscrape.com/ and get the HMTL text from the homepage.**

```python
url = 'http://quotes.toscrape.com/'
response = requests.get(url)
soup = BeautifulSoup(response.text, "html.parser")
qoutes =soup.select('.text')
for q in qoutes:
  print(q.text)
```

**TASK: Get the names of all the authors on the first page.**

```python
authors = soup.select('.author')
for a in authors:
  print(a.text)
```

**TASK: Inspect the site and use Beautiful Soup to extract the top ten tags from the requests text shown on the top right from the home page (e.g Love,Inspirational,Life, etc...). HINT: Keep in mind there are also tags underneath each quote, try to find a class only present in the top right tags, perhaps check the span.**

```python
import requests
from bs4 import BeautifulSoup
response =requests.get("http://quotes.toscrape.com/")
soup = BeautifulSoup(response.text,"html.parser")
tags = soup.select('.tag-item')
for t in tags:
  print(t.text)
```

**TASK: Notice how there is more than one page, and subsequent pages look like this http://quotes.toscrape.com/page/2/. Use what you know about for loops and string concatenation to loop through all the pages and get all the unique authors on the website.

```python
import requests
from bs4 import BeautifulSoup
base_url = "http://quotes.toscrape.com/"
page_url =base_url
authors =set()
for page in range(1,11):
  url = base_url.format(page)
  response = requests.get(url)
  soup = BeautifulSoup(response.text,"html.parser")
  for a in soup.select('.author'):
    authors.add(a.text)
print(authors)
print("total unique authors:",len(authors))
```
